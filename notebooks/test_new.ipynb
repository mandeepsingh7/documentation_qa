```python
import requests 
from scrapy.http import HtmlResponse
```

```python
url = "https://docs.nvidia.com/cuda/cuda-runtime-api/driver-vs-runtime-api.html#driver-vs-runtime-api"
```

```python
r = requests.get(url)
```

```python
response = HtmlResponse(url=url, body=r.text, encoding='utf-8')
```

```python
response.css("title::text").get()
```

```python
response.css("article#contents h2::text").getall()
```

```python
response.css("span.member_name ::text").getall()
```

```python
response.css("article#contents h2::text").getall()
```

```python
texts = response.xpath('//article[@id="contents"]//text()').getall()

cleaned = [t.strip() for t in texts if t.strip()]

print("\n".join(cleaned))
```

```python
texts = response.css("article#contents ::text").getall()

cleaned = [t.strip() for t in texts if t.strip()]

print("\n".join(cleaned))
```
